# import the modules

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent 
from langchain_core.tools import tool
from groq import Groq
```

# explore models available in Groq

```python

client = Groq()   # يقرأ GROQ_API_KEY من البيئة
for m in client.models.list().data:
    print(m.id)
```

#  initialize LLM


```python
# initialize LLM
llm = ChatGroq(
    # model name
    model="openai/gpt-oss-20b",
    # randomness
    temperature=0.1,
)
```

# create a tool

```python
# 2. Create a tool
@tool 
#input is a string
#output is a string
def get_weather(location: str) -> str:
    #docstring is used by the agent to understand which tool to call
    """Get current weather information for a specific location.

    Args:
        location: The city or location to get weather for

    Returns:
        A string describing the current weather conditions, or a message if data is not available
    """
    # Mock implementation - in production, call a real weather API
    weather_data = {
        "new york": "Sunny, 72°F",
        "london": "Cloudy, 59°F",
        "tokyo": "Clear, 68°F",
    }
#strip any whitesapces 
    location_key = location.lower().strip()

    if location_key in weather_data:
        return f"Weather in {location}: {weather_data[location_key]}"
    else:
        return f"Weather data not available for {location}"

```

# create the agent 

```python
# 3. Create the agent with the model and the tool
agent = create_agent(llm, tools=[get_weather])
```

# Helper Function: ask_agent

```python
def ask_agent(question: str):
    """Helper function to ask the agent a question and print the answer."""
    # messages : history 
    # Invoke the agent : role user : the human , role assistant : the ai , role system : the system role, content : the question
    result = agent.invoke({
        "messages": [{"role": "user", "content": question}]
    })

    # Extract the final message
    final_message = result["messages"][-1]
    print(final_message.content)
```

# try ask the agent 

```python
ask_agent("What is the weather in New York?")
```
